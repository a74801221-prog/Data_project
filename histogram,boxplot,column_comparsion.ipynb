```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('student_data.csv')
sns.set_theme(style='whitegrid')
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
```

```python
df[numerical_cols].hist(bins=15, figsize=(15, 10), color='teal', grid=False)
plt.suptitle(
    'Histogram Distribution for All Numerical Columns',
    fontsize=16,
    fontweight='bold',
)
plt.savefig('all_histograms.png', dpi=300, bbox_inches='tight')
plt.show()
print('--------------------------------------------------')
print('>> [Analatical Note - Histogram]:')
print(
    'The frequency plots illustrate the distribution of values across'
    ' numerical columns; for instance, student ages are concentrated in a'
    ' specific range, whereas distribution patterns and spreads vary'
    ' significantly in student grades and absences.'
)
print('--------------------------------------------------\n')
```

```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=df[numerical_cols], palette='Set2')
plt.title(
    'Box Plot for Outliers Detection (All Numerical Columns)',
    fontsize=14,
    fontweight='bold',
)
plt.xticks(rotation=45)
plt.savefig('all_boxplots.png', dpi=300, bbox_inches='tight')
plt.show()
print('>> Analytical Note - Box plot: The box plot displays the median and IQR for each column, and extreme points appear as outliers outside the whiskers, especially in absences and grades, reflecting individual cases that require attention.')
```

```python
# اختيار أهم الأعمدة الرقمية للمقارنة
important_cols = ['absences', 'G1', 'G2', 'G3']

# رسم مقارنة تفصيلية بين الأعمدة المهمة
plt.figure(figsize=(10, 6))
sns.pairplot(df[important_cols], diag_kind='kde')
plt.suptitle('Comparison Between Key Numerical Columns', y=1.02, fontsize=14, fontweight='bold')
plt.savefig('important_columns_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# طباعة الملاحظة التحليلية الخاصة بالمقارنة
print(">> Analytical Note - Comparison: Comparing key numerical columns like grades (G1, G2, G3) shows a strong positive correlation, while absences reveal how higher values can impact overall student performance.")
```
