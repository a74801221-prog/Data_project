```python
import pandas as pd


missing_data = pd.DataFrame({
    'عدد القيم المفقودة': df.isnull().sum(),
    'النسبة المئوية (%)': (df.isnull().sum() / len(df)) * 100
})

missing_data = missing_data.sort_values(by='عدد القيم المفقودة', ascending=False)

display(missing_data)
```
