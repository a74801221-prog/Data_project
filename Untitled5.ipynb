```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("student_data.csv")

print("First 5 rows of the dataset:")
display(df.head())

numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(12, 10))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    cbar=True,
)
plt.title(
    "Correlation Heatmap for Student Dataset", fontsize=16, fontweight="bold"
)
plt.show()


plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df, x="G1", y="G3", hue="sex", palette="Set1", alpha=0.8, s=70
)
plt.title(
    "Relationship between First Period Grade (G1) and Final Grade (G3)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("G1 (First Period Grade)")
plt.ylabel("G3 (Final Grade)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()


plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="studytime",
    y="G3",
    hue="address",
    palette="viridis",
    alpha=0.8,
    s=80,
)
plt.title(
    "Relationship between Study Time and Final Grade (G3)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Study Time (1: <2 hrs, 2: 2-5 hrs, 3: 5-10 hrs, 4: >10 hrs)")
plt.ylabel("G3 (Final Grade)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
