```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# قراءة البيانات
df = pd.read_csv('student_data.csv')

# إنشاء مجلد لحفظ الصور الناتجة تلقائياً
if not os.path.exists('final_charts'):
    os.makedirs('final_charts')

print("تم تحميل البيانات بنجاح، عدد الصفوف:", len(df))
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# رسم Bar Chart لمقارنة متوسط G3 حسب توفر الإنترنت
ax = sns.barplot(x="internet", y="G3", data=df, palette="Purples_d", ci=None)

plt.title('Comparison of Final Grade (G3) by Internet Access', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Internet Access at Home', fontsize=12)
plt.ylabel('Average Final Grade (G3)', fontsize=12)

# تحسين مظهر القيم على الأعمدة
for p in ax.patches:
    ax.annotate(f"{p.get_height():.2f}", 
                (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', fontsize=11, color='white', fontweight='bold')

plt.tight_layout()
plt.savefig('final_charts/bar_chart_internet_g3.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# رسم Count Plot لمقارنة أعداد الطلاب في المدارس حسب الجنس
ax = sns.countplot(x="school", hue="sex", data=df, palette="Set2")

plt.title('Student Count by School and Gender', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('School (GP: Gabriel Pereira, MS: Mousinho da Silveira)', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.legend(title='Gender')

# إضافة الأعداد فوق الأعمدة
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{int(height)}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.savefig('final_charts/count_plot_school_sex.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# رسم Box Plot لاكتشاف القيم الشاذة وتوزيع الدرجات
sns.boxplot(x="goout", y="G3", data=df, palette="coolwarm")

plt.title('Final Grade (G3) Distribution by Going Out Frequency', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Going Out with Friends (1: Very Low, 5: Very High)', fontsize=12)
plt.ylabel('Final Grade (G3)', fontsize=12)

plt.tight_layout()
plt.savefig('final_charts/box_plot_goout_g3.png', dpi=300)
plt.show()
```
